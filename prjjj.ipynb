```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats, integrate
import sympy as sp


print("Arithmetic, Calculus, Probability (functions which need exponents use ** instead of ^)")
choice = input()
if choice == "Arithmetic":
    print("Enter two numbers:")
    num1 = float(input())
    num2 = float(input())
    print("Choose operation: +, -, *, /")
    operation = input()
    if operation == "+":
        result = num1 + num2
    elif operation == "-":
        result = num1 - num2
    elif operation == "*":
        result = num1 * num2
    elif operation == "/":
        result = num1 / num2
    else:
        print("Invalid operation")
        exit()
    print("Result: ", result)


if choice == "Calculus":
    print("Integral or Derivatives")
    if input() == "Integral":
        print("Definite or Indefinite or Double (only over unit square)")
        thing = input()
        if thing == "Definite":
            print("Enter function:")
            func = input()
            x = sp.symbols('x')
            lower = float(input("Enter lower limit: "))
            upper = float(input("Enter upper limit: "))
            integral1 = sp.integrate(func, (x, lower, upper))
            display(integral1)
        elif thing == "Indefinite":
            print("Enter function:")
            func = input()
            x = sp.symbols('x')
            integral = sp.integrate(func,x)
            display(integral)
        elif thing == "Double":
            print("Enter function:")
            func = input()
            x, y = sp.symbols('x y')
            integral2 = sp.integrate(x * y, (x, 0, 1), (y, 0, 1))
            display(integral2)
    if input() == "Derivatives":
        print("Enter function:")
        func = input()
        x = sp.symbols('x')
        derivative = sp.diff(func,x)
        display(derivative)

if choice == "Probability":
    print("Enter PDF of a continous random variable:")
    func1 = input()
    x = sp.symbols('x')
    func1 = sp.sympify(func1)
    mean = sp.integrate(x * func1, (x, 0, 1))
    var = sp.integrate(x**2 * func1, (x, 0, 1)) - mean**2
    print(f"E[X] = {mean},  Var(X) = {var})")
```
